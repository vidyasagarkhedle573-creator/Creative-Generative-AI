Q1. Explain how generative AI is used in creative domains: poetry, storytelling,
music, art and video. For each domain name one kind of model or tool, what input it
takes and what it produces. Explain the difference between AI as a tool that assists a
creator and AI that replaces a creator.

Generative AI creates new content by learning patterns from existing examples.

1. **Poetry:** A text-generation language model takes a topic, keywords or a prompt as input and produces lines of poetry.
2. **Storytelling:** A language model takes characters, a setting or a story idea as input and produces a story with events and dialogue.
3. **Music:** A music-generation model takes a text description, melody or MIDI notes as input and produces a musical sequence or an audio track.
4. **Art:** A text-to-image diffusion model takes a written description as input and produces an image.
5. **Video:** A text-to-video or image-to-video model takes a text prompt or an image as input and produces a short video.

When AI assists a creator, it helps with ideas, drafts, editing or repetitive work. The human creator selects, changes and approves the final result. When AI replaces a creator, the system produces most or all of the content with little human involvement. AI assistance supports human decisions, while replacement gives more of the creative work to the system.

Q2. Explain how a language model generates a poem or a story: next-token
prediction, temperature, top-p and the role of the prompt. Write three different
prompts for a short story about the same idea (a neutral prompt, a style-guided
prompt and a constraint-based prompt) and explain how the outputs would differ. Why
do high temperature settings make writing more creative but less consistent?

A language model generates text one token at a time. A token may be a word or part of a word. It calculates probabilities for possible next tokens and selects one. This is called **next-token prediction**. The selected token is added to the text, and the process repeats.

**Temperature** controls how strongly the model follows the most likely tokens. A low temperature usually gives safer and more predictable writing. A high temperature gives less likely choices a better chance.

**Top-p** sampling selects from the smallest group of likely tokens whose combined probability reaches the chosen p value. It limits the choices available at each step.

The **prompt** tells the model the topic, purpose, style and other requirements.

Three prompts for the same idea, a lost dog finding its way home:

- **Neutral prompt:** “Write a short story about a lost dog that finds its way home.”
- **Style-guided prompt:** “Write a warm and emotional short story about a lost dog finding its way home, using simple language and vivid descriptions.”
- **Constraint-based prompt:** “Write a story of exactly 150 words about a lost dog finding its way home. Use three paragraphs and include a helpful shopkeeper.”

The neutral prompt allows the model to choose the style and structure. The style-guided prompt encourages emotional and descriptive writing. The constraint-based prompt follows specific length and structure requirements.

High temperature can make writing more creative because the model may choose less common words and ideas. However, those choices may not fit the earlier events or meaning, so the writing can become less consistent.

Q3. Explain how AI music generation works at a high level: representing music
as symbols (MIDI) or as audio, sequence models, and diffusion or transformer models
for audio. List three challenges of generating music (long-term structure, quality,
rights over training data) and one way to handle each.

AI music generation can represent music in two main ways. **MIDI** represents musical events such as notes, pitch, duration, timing and instruments. An audio representation contains the sound itself, often as a waveform or a compressed audio representation.

A sequence model learns the order of musical events. It can predict the next note or musical unit from the earlier sequence. Transformer models use attention to learn relationships between different parts of a sequence. Diffusion models start with noise and gradually produce a clearer audio result. Audio transformers can also generate audio units in sequence.

Three challenges and possible solutions are:

1. **Long-term structure:** A generated song may lose its theme or arrangement over time. A possible solution is to generate the song in sections using an overall structure plan.
2. **Audio quality:** The output may contain noise, unnatural instruments or timing problems. A possible solution is to use high-quality training data and review the generated audio.
3. **Rights over training data:** Training music may be used without permission. A possible solution is to use licensed or permission-based datasets and respect creator opt-outs.

Q4. Write a Python program that uses a very simple Markov chain (a dictionary
of word to next-word lists) built from a short paragraph of your choice to generate a
20-word piece of text, with a fixed random seed. Explain how this toy model is
different from a large language model and why its output is less coherent.

The following program builds a dictionary in which each word points to the words that followed it in the paragraph. It then chooses the next word randomly. The fixed seed makes the result repeatable.

````python
import random
import re

paragraph = """the little bird sat on the tree. the bird sang a song.
the little bird flew over the garden. the garden had a tall tree."""

words = re.findall(r"[a-z]+", paragraph.lower())

markov = {}
for current_word, next_word in zip(words, words[1:]):
    markov.setdefault(current_word, []).append(next_word)

random.seed(10)

current = "the"
generated = [current]

for _ in range(19):
    next_words = markov.get(current)
    if not next_words:
        current = random.choice(words)
        generated.append(current)
    else:
        current = random.choice(next_words)
        generated.append(current)

print(" ".join(generated))
````

**Output:**

```text
the garden the little bird flew over the little bird sang a song the bird sat on the little bird
```

This toy model only remembers which words appeared next to each other in the paragraph. It does not understand meaning, grammar or the topic. It also has a very small amount of training text.

A large language model is trained on a much larger collection of text and uses a neural network with many parameters to learn complex language patterns and relationships. Therefore, it can usually produce more relevant and coherent text. The Markov chain may repeat words or move between unrelated parts of the paragraph, so its output is less coherent.

Q5. Explain the ethical issues of creative generative AI: copyright and training
data, originality and credit to artists, plagiarism, deepfakes and misinformation, and
the effect on creative jobs. Give one real or realistic example for each issue and one
possible response (for example licensing, watermarking, disclosure, opt-out).

Creative generative AI raises several ethical issues.

1. **Copyright and training data:** A model may be trained on artwork or music without the creator's permission. For example, an artist's online illustrations may be included in a training dataset. A response is to use licensed datasets, obtain permission and provide an opt-out process.
2. **Originality and credit:** AI-generated work may closely resemble the style or ideas of an existing artist. For example, a generated poster may imitate a living illustrator's distinctive work. A response is to disclose AI use and credit human contributors where appropriate.
3. **Plagiarism:** A model may produce text or images that are too similar to an existing work. For example, a student may submit generated text that copies parts of an online article. A response is to check outputs for similarity and cite original sources.
4. **Deepfakes and misinformation:** AI can create realistic but false images, audio or videos. For example, a fake video may show a public figure making a statement they never made. A response is to label synthetic media, use provenance tools and verify content before sharing.
5. **Effect on creative jobs:** Some routine creative tasks may be automated, changing the work available to designers, writers and editors. For example, a company may use AI for basic product images instead of commissioning every image. A response is to train workers in AI-assisted workflows and keep human creators involved in review and original work.

Q6. Explain responsible-use practices for generative AI content: disclosure that
content is AI-generated, watermarking and content provenance, human review, bias
checks and safe-content filters. Write a short policy of at least eight rules for a
company that uses generative AI to write marketing content.

Responsible use means making AI content transparent, checking it before publication and protecting people affected by it.

**Company policy for AI-generated marketing content**

1. Disclose AI-generated or substantially AI-assisted content when it could affect the audience's understanding.
2. Do not use AI to create false product claims, fake reviews or misleading endorsements.
3. Use licensed, approved or permission-based material as input.
4. Add suitable watermarks or provenance information to synthetic images, audio and video where supported.
5. Require a human employee to review and approve content before publication.
6. Check content for stereotypes, unfair bias and discriminatory language.
7. Use safe-content filters to prevent harmful, abusive or inappropriate material.
8. Do not enter customers' personal or confidential information into unapproved AI tools.
9. Check factual claims, prices, offers and product details against official company information.
10. Keep a record of the AI tool used and the person who approved the final content.

Q8. Explain how reinforcement learning is used with generative models: reward
signals, reinforcement learning from human feedback (RLHF), and reward design for
creative tasks (how do you reward a good poem?). Explain reward hacking with an
example and why creative quality is hard to measure automatically.

Reinforcement learning improves a model by using feedback called a **reward signal**. The model produces an answer or other output, and a reward indicates how well that output meets the goal. The model is adjusted to produce outputs that receive better rewards.

**Reinforcement Learning from Human Feedback (RLHF)** uses human preferences to guide the model. People compare or rate model responses. This feedback is used to train a reward model, which estimates how preferred an output is. Reinforcement learning then uses that reward model to improve the language model's responses.

For a creative task such as writing a poem, reward design could consider relevance to the topic, clear language, rhythm, originality and human ratings. Human reviewers can compare poems and select the ones they prefer.

**Reward hacking** happens when a model finds a way to get a high reward without achieving the real goal. For example, if a poem is rewarded mainly for using rhyming words, the model may produce many rhymes that do not make sense.

Creative quality is difficult to measure automatically because people have different tastes. A poem can be meaningful to one reader but not to another. Creativity, emotion and originality are also difficult to represent with a single score.

Q9. Explain multimodal systems: models that handle text, images, audio and
video together. Explain the idea of a shared embedding space and cross-modal learning
(for example text-to-image, image captioning, speech-to-text and text-to-speech). Draw
a block diagram of a system that takes an image and a question and produces a spoken
answer.

A multimodal system is an AI system that can process or generate more than one type of information, such as text, images, audio and video.

A **shared embedding space** represents different types of data as numerical vectors in a common space. Related information, such as an image of a dog and the words “a dog”, can have similar representations. This helps a model connect information from different modes.

**Cross-modal learning** means learning relationships between different data types. Examples include:
- **Text-to-image:** Uses a text description to generate an image.
- **Image captioning:** Uses an image to generate a text description.
- **Speech-to-text:** Converts spoken audio into written words.
- **Text-to-speech:** Converts written text into spoken audio.
